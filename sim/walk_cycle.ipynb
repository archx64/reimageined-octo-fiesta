```python
from matplotlib import pyplot as plt

f = 4 # four cycles per second
duration =  1 # seconds
fs = 1000 # samples per second (sampling rate)

t = np.linspace(0, duration, int(fs * duration), endpoint=False)

x = 2 * np.pi * f * t # how far along the sine curve the point is at each time

y = np.sin(x) - 0.5
z = np.cos(x)

plt.figure(figsize=(5,5))
plt.grid(True)
plt.plot(t, y)
plt.xlabel('time (s)')
plt.ylabel('height of the point')
plt.title('A point moving at 4 cycles per second')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

f = 1                # cycles per second
duration = 2.0             # seconds of animation
fps = 60                   # frames per second

# Draw the full path the point will travel: f cycles/s × duration seconds
x_path = np.linspace(0, 2 * np.pi * f * duration, 2000)

fig, ax = plt.subplots()
ax.plot(x_path, np.sin(x_path), color="lightgray")
(point,) = ax.plot([], [], "o", color="red", markersize=10)
ax.set_xlabel("x (radians)")
ax.set_ylabel("sin(x)")

def update(frame):
    t = frame / fps                    # elapsed time in seconds
    x = 2 * np.pi * f * t              # position along the curve
    point.set_data([x], [np.sin(x)])   # set_data needs lists, not bare numbers
    return (point,)

anim = FuncAnimation(fig, update, frames=int(duration * fps),
                     interval=1000 / fps, blit=True)
plt.show()
```

```python
from IPython.display import HTML
HTML(anim.to_jshtml())                       # interactive player inside the notebook

anim.save("sine.gif", writer="pillow", fps=fps) 
```
